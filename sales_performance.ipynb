# Data Cleaning

## 1. Load the Dataset

```python
import pandas as pd

df = pd.read_csv("Sample-Superstore.csv",encoding="latin1")
df.head()
```

## 2. Handle null and duplicates

```python
df.isnull().sum()
```

```python
print(int(df.duplicated().sum()))
```

```python
df=df.drop_duplicates()
```

```python
print(int(df.duplicated().sum()))
```

## 3. Data type conversion

```python
df.info()
```

### As we can see order date and ship date are in object type---> date type

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])
```

# Data transformation

## 1. Normalization

```python
from sklearn.preprocessing import MinMaxScaler

## Min max normalization

scaler = MinMaxScaler()
df[['Sales_Normalized', 'Profit_Normalized']] = scaler.fit_transform(df[['Sales', 'Profit']])
```

```python
df.info()
```

### Feature Engineering

```python
# Extract time features
df['Year'] = df['Order Date'].dt.year
df['Month'] = df['Order Date'].dt.month
df['Month_Name'] = df['Order Date'].dt.month_name()

# Profit margin
df['Profit_Margin'] = df['Profit'] / df['Sales']
```

```python
# Pareto analysis
product_sales = df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False)
product_sales_cum = product_sales.cumsum() / product_sales.sum()
```

```python
df.info()
```

```python
columns_to_drop = [
    'Row ID', 'Order ID', 'Customer ID', 'Customer Name',
    'Product ID', 'Country', 'Ship Date', 'Ship Mode', 'Postal Code'
]
df_final = df.drop(columns=columns_to_drop)
df_final.head()
```

```python
df_final.to_csv("final_superstore_dashboard.csv", index=False)
```
